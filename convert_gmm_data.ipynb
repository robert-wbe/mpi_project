```python
import torch
```

```python
gmms = torch.load('trained_checkpoints/celeba_nested_nested_gmm_np.pt')
```

```python
gmms[0].covariances_.shape
```

```python
subweights = torch.stack([
    torch.as_tensor(
        gmm.weights_, dtype=torch.float32, device='cuda'
    ) for gmm in gmms
])
```

```python
means = torch.stack([
    torch.as_tensor(
        gmm.means_, dtype=torch.float32, device='cuda'
    ) for gmm in gmms
])
```

```python
variances = torch.stack([
    torch.as_tensor(
        gmm.covariances_, dtype=torch.float32, device='cuda'
    ) for gmm in gmms
])
```

```python
supweights = torch.load('trained_checkpoints/celeba_stablediff_nested_gmm.pt')['supweights']
```

```python
torch.save({
    'supweights' : supweights,
    'subweights': subweights,
    'means': means,
    'variances': variances,
}, 'trained_checkpoints/celeba_nested_nested_gmm.pt')
```
